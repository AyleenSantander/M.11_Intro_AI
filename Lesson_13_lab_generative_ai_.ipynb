## Define a function for text generation
This function will be our main tool for generating text. It takes a prompt and several parameters that control the generation process.

```python
def generate_text(prompt, max_length=100, temperature=1.0, top_k=50, top_p=0.95):
    input_ids = tokenizer.encode(prompt, return_tensors="pt")
    
    output = model.generate(
        input_ids,
        max_length=max_length,
        temperature=temperature,
        top_k=top_k,
        top_p=top_p,
        do_sample=True,
        num_return_sequences=1
    )
    
    generated_text = tokenizer.decode(output[0], skip_special_tokens=True)
    return generated_text
```

### Exercise 4: Using top_k and top_p
Here, we'll experiment with the `top_k` and `top_p` parameters to see how they affect text generation.

`top_k` and `top_p` are both parameters used in text generation to control the randomness and quality of the output. They work by limiting the pool of words the model can choose from at each step of the generation process:

1. `top_k`:
   - Selects the k most likely next words.
   - The model then only chooses from this reduced set of words.
   - A lower k value makes the output more focused but potentially less creative.
   - Example: If top_k = 50, only the 50 most probable next words are considered.

2. `top_p` (also known as nucleus sampling):
   - Selects the smallest set of words whose cumulative probability exceeds p.
   - More dynamic than top_k, as the number of words can vary.
   - Helps maintain diversity while avoiding low-probability words.
   - Example: If top_p = 0.9, it selects the most probable words until their cumulative probability reaches 90%.

In essence:
- top_k limits the absolute number of words considered.
- top_p limits based on cumulative probability.

Both parameters help strike a balance between coherence and creativity in generated text. They're often used together, with top_p typically applied after top_k filtering.

### **More Examples**

````python


def ask_gpt2(prompt, tokenizer, model, max_length=100, temperature=0.7):
    """
    This function sends a prompt to the GPT-2 model and returns the generated response.
    It simulates a chat-like interaction similar to the provided ChatGPT function.

    Parameters:
    - prompt (str): The input text prompt/question for GPT-2.
    - max_length (int): The maximum length of the generated response.
    - temperature (float): Controls randomness in the output. Higher values make the output more random.

    Returns:
    - str: The generated response from GPT-2.
    """
    try:
        # Tokenize the prompt
        input_ids = tokenizer.encode(prompt, return_tensors="pt")

        # Generate a response
        output = model.generate(
            input_ids,
            max_length=max_length + len(input_ids[0]),
            temperature=temperature,
            num_return_sequences=1,
            pad_token_id=tokenizer.eos_token_id,
            do_sample=True
        )

        # Decode the response
        response = tokenizer.decode(output[0], skip_special_tokens=True)

        # Remove the original prompt from the response
        response = response[len(prompt):].strip()

        return response
    except Exception as e:
        return f"Error: {e}"

# Example usage
system_prompt = "You are a helpful assistant."
user_prompt = "What are the three laws of robotics?"

full_prompt = f"{system_prompt}\nUser: {user_prompt}\nAssistant:"
response = ask_gpt2(full_prompt, tokenizer, model, max_length=150) # Pass the 'model' argument
print("User:", user_prompt)
print("Assistant:", response)

````

````python
# Example: Simulate a conversation with ChatGPT
def chat_with_gpt():
    """
    This function allows the user to have a multi-turn conversation with ChatGPT.
    """
    print("Start chatting with ChatGPT! Type 'exit' to end the chat.\n")
    while True:
        user_input = input("You: ")
        if user_input.lower() == 'exit':
            print("Ending chat. Goodbye!")
            break
        response = ask_gpt2(user_input)
        print("ChatGPT:", response)

# Uncomment the line below to start a conversation with ChatGPT in the Colab environment
# chat_with_gpt()
````

````python
# Call ask_gpt2 with tokenizer and model
prompt = """
Write a blog post about the benefits of learning Python for data science.
The post should cover:
- Why Python is important for data science.
- Key libraries and tools to use.
- Real-world applications in various industries.
"""
response = ask_gpt2(prompt, tokenizer, model) # Pass tokenizer and model
print("Generated Blog Post:\n", response)
````
