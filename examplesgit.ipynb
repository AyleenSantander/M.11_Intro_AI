```python
path = '/content/drive/MyDrive/Colab_Notebooks/Lesson_4_lab_data_processing.csv'

with open(path) as f:
    lines = f.readlines()

print(f"Total lines: {len(lines)}\n")

print("--- FIRST 15 LINES ---")
for i, line in enumerate(lines[:15]):
    fields = line.rstrip('\n').split(',')
    print(f"[{i}] ({len(fields)} fields)  {line.rstrip()}")

print("\n--- LINES 80-90 ---")
for i, line in enumerate(lines[80:90], start=80):
    fields = line.rstrip('\n').split(',')
    print(f"[{i}] ({len(fields)} fields)  {line.rstrip()}")
```

```python
path = '/content/drive/MyDrive/Colab_Notebooks/Lesson_4_lab_data_processing.csv'

with open(path, 'r', encoding='utf-8', errors='replace') as f:
    lines = f.readlines()

print('Total lines:', len(lines))
print()

for i, line in enumerate(lines[:15]):
    n = len(line.rstrip('\n').split(','))
    print(i, '| fields:', n, '|', line.rstrip())

print()

for i in range(80, min(90, len(lines))):
    n = len(lines[i].rstrip('\n').split(','))
    print(i, '| fields:', n, '|', lines[i].rstrip())
```
