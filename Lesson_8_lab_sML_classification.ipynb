The three plots you'd typically use:

msno.bar(df) — a bar for each column showing how many values are present. Quick answer to "how bad is it, and where?"

msno.matrix(df) — the useful one. Each row of your dataframe is a thin horizontal line, white where a value is missing. This shows whether missingness is scattered randomly or clustered in blocks. Blocks mean something systematic happened — a sensor failed, a form question was added partway through data collection, two datasets got concatenated.

msno.heatmap(df) — correlation between missingness across columns. If Insulin and SkinThickness are missing together in the same rows, that tells you both came from the same skipped measurement, not two independent problems.

Why it matters for your work: the pattern determines how you handle it. If values are missing at random, imputing with a median is defensible. If they're missing in a systematic block tied to something else in the data, imputation will bake a bias into your model, and dropping the column may be the better call.

On the Pima dataset specifically, the heatmap will show Insulin and SkinThickness missing together — both are harder tests, and it looks like they were often skipped on the same patients. That's a real finding worth a sentence in your notebook rather than just silently filling with medians. -->

Notes on Scaling Data

Scaling data is like adjusting measurements so you can compare things fairly. For example, comparing someone's height in centimeters to their shoe size wouldn't make sense. Scaling ensures all features (e.g., height, weight) contribute equally to a machine learning model's decision.

Two common methods:

1. **Normalization:** Squishes data into a 0 to 1 range (like grading on a curve).
2. **Standardization:** Z-score Normalisation; Centers data around zero based on how spread out it is (like comparing everyone's height relative to the average).

Use scaling for:

* Algorithms that measure distance between data points.
* Models that learn by adjusting themselves (gradient descent).

Key points:

* Scale training and test data separately.
* Don't scale categories (like eye color).

It's like coaching a basketball team: Scaling helps you utilize each player's strengths effectively, regardless of their initial differences. This improves overall team performance, just as scaling improves a machine learning model's performance.

## **Test Train Split and Cross Validation methods**



***Train Test Split*** : To have unknown datapoints to test the data rather than testing with the same points with which the model was trained. This helps capture the model performance much better.

![](https://cdn-images-1.medium.com/max/1600/1*-8_kogvwmL1H6ooN1A1tsQ.png)

***Cross Validation***: When model is split into training and testing it can be possible that specific type of data point may go entirely into either training or testing portion. This would lead the model to perform poorly. Hence over-fitting and underfitting problems can be well avoided with cross validation techniques

![](https://cdn-images-1.medium.com/max/1600/1*4G__SV580CxFj78o9yUXuQ.png)


***About Stratify*** : Stratify parameter makes a split so that the proportion of values in the sample produced will be the same as the proportion of values provided to parameter stratify.

For example, if variable y is a binary categorical variable with values 0 and 1 and there are 25% of zeros and 75% of ones, stratify=y will make sure that your random split has 25% of 0's and 75% of 1's.

For Reference : https://towardsdatascience.com/train-test-split-and-cross-validation-in-python-80b61beca4b6

```python
#importing train_test_split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=1/3,random_state=42, stratify=y)
```

### **Create a classifier model** (You can try other algorithm after this example)

- In the k-nearest neighbors (kNN) classifier, the "k" represents the number of nearest neighbors to consider when making a prediction for a new data point.

- Choosing the Right "k"

  The value of k is a hyperparameter that you need to choose.

  There's no one-size-fits-all answer, as the optimal k depends on your dataset and problem.

   -  **Small k**: A smaller k makes the model more sensitive to noise and outliers, potentially leading to overfitting.
   - **Large k**: A larger k makes the model more resistant to noise, but it might oversimplify the decision boundaries and underfit.

- Finding the Best k

  You can **experiment** with different values of k and evaluate the model's performance using techniques like cross-validation. The goal is to find the k that gives you the best balance between overfitting and underfitting, leading to the best possible accuracy on new, unseen data.





```python
#Fix the error before replace 0s instead Nan
train_scores, test_scores = [], []
neighbors = range(1, 21)

for k in neighbors:
    knn = Pipeline([
        ('imputer', SimpleImputer(strategy='median')),
        ('scaler', StandardScaler()),
        ('model', KNeighborsClassifier(n_neighbors=k))
    ])
    knn.fit(X_train, y_train)
    train_scores.append(knn.score(X_train, y_train))
    test_scores.append(knn.score(X_test, y_test))
```

#### **Experiment with visualisation to get the most suitable k for Model training**

```python
# choosing k btw 1 and 15
test_scores = []
train_scores = []

for i in range(1,15):

    knn = KNeighborsClassifier(i)
    knn.fit(X_train,y_train)

    train_scores.append(knn.score(X_train,y_train))
    test_scores.append(knn.score(X_test,y_test))
```

```python
## score that comes from testing on the same datapoints that were used for training
max_train_score = max(train_scores)
train_scores_ind = [i for i, v in enumerate(train_scores) if v == max_train_score]
print('Max train score {} % and k = {}'.format(max_train_score*100,list(map(lambda x: x+1, train_scores_ind))))
```

```python
## score that comes from testing on the datapoints that were split in the beginning to be used for testing solely
max_test_score = max(test_scores)
test_scores_ind = [i for i, v in enumerate(test_scores) if v == max_test_score]
print('Max test score {} % and k = {}'.format(max_test_score*100,list(map(lambda x: x+1, test_scores_ind))))
```

small k means the model follows every little wiggle in the data including the noise (overfitting). Large k means it smooths everything out and misses real patterns (underfitting). You pick the middle.

```python
#Cross-Validation basic version
pipe = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler()),
    ('model', KNeighborsClassifier(n_neighbors=11))
])

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
scores = cross_val_score(pipe, X, y, cv=cv, scoring='accuracy')

print(scores)
print(f"Mean: {scores.mean():.4f}  Std: {scores.std():.4f}")
```

StratifiedKFold keeps the class balance the same in every fold. That matters here — the Pima data is about 65/35, so a plain KFold could hand you a fold with an unrepresentative mix.
instead of trusting one lucky split, you try several and average. More honest estimate.

```python
#Finding the best k with cross-validation:
k_range = range(1, 31)
cv_scores = []

for k in k_range:
    pipe.set_params(model__n_neighbors=k)
    cv_scores.append(cross_val_score(pipe, X, y, cv=cv).mean())

best_k = k_range[int(np.argmax(cv_scores))]
print(f"Best k: {best_k}  ({max(cv_scores):.4f})")

plt.figure(figsize=(10, 5))
plt.plot(k_range, cv_scores, marker='o')
plt.xlabel('k')
plt.ylabel('CV Accuracy')
plt.show()
```

```python
#GridSearchCV, which is the version worth putting in a portfolio:
from sklearn.model_selection import GridSearchCV

param_grid = {
    'model__n_neighbors': range(1, 31), #30 options
    'model__weights': ['uniform', 'distance'], #2 options
    'model__metric': ['euclidean', 'manhattan'] #2 options
}

grid = GridSearchCV(pipe, param_grid, cv=cv, scoring='accuracy', n_jobs=-1)
grid.fit(X, y)

print(grid.best_params_)
print(f"{grid.best_score_:.4f}")
```

k = 25, a fairly large neighbourhood. The model is averaging over 25 similar patients before deciding, which means it's smoothing rather than chasing noise. Sensible for a dataset this small and this messy.

manhattan, distance measured by summing the differences along each axis, rather than straight-line (euclidean). Manhattan often wins on higher-dimensional data because euclidean distances start to bunch together as dimensions grow, making everything look equally far apart.

uniform, all 25 neighbours get an equal vote. The alternative, distance, would weight closer ones more heavily. That it lost suggests there's no tight local structure worth exploiting here.

0.7604, 76% average accuracy across your 5 folds.

Is that good? Put it against the baseline: predicting "no diabetes" for everyone scores about 65% on this dataset. So you're 11 points above doing nothing. That's real but modest, and it's roughly what people get on Pima, the ceiling is around 77–79%. The data has 768 rows and heavily imputed columns, so there's a limit to what any model can extract.

```python
#Check what the accuracy is hiding:
y_pred = grid.predict(X_test)
print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))
print(f"ROC-AUC: {roc_auc_score(y_test, grid.predict_proba(X_test)[:, 1]):.4f}")
```

The headline problem: 36 of 89 diabetic patients were classified as healthy. That's recall of 0.60, the model misses 40% of actual cases. Meanwhile it handles the healthy class at 0.89 recall. Accuracy of 0.79 papers over that gap because the healthy group is nearly twice as large.
ROC-AUC of 0.8548 is solid. This mean the model's ranking is good, it does assign higher probabilities to diabetics on the whole. The problem isn't the model, it's the 0.5 decision threshold. 

```python
probs = grid.predict_proba(X_test)[:, 1]

for t in [0.5, 0.4, 0.35, 0.3, 0.25]:
    pred = (probs >= t).astype(int)
    print(f"threshold {t}")
    print(confusion_matrix(y_test, pred))
    print(classification_report(y_test, pred, digits=2))
```

At around 0.3 you'll likely pull recall on class 1 up to 0.75–0.80, at the cost of more false positives. That's the trade you want here.

The other lever is re-running the grid search with scoring='recall' or scoring='f1' instead of accuracy, so the parameter selection optimizes for the thing you care about rather than overall correctness.

Accuracy 0.79 but recall on the positive class only 0.60, so threshold tuning was applied" is the kind of observation that reads as analytical maturity rather than just running the default pipeline.

### Classification Report

Report which includes Precision, Recall and F1-Score.


#### Precision Score
        TP – True Positives
        FP – False Positives

        Precision – Accuracy of positive predictions.
        Precision = TP/(TP + FP)
        
   
#### Recall Score
        FN – False Negatives

        Recall(sensitivity or true positive rate): Fraction of positives that were correctly identified.
        Recall = TP/(TP+FN)
        
#### F1 Score
        F1 Score (aka F-Score or F-Measure) – A helpful metric for comparing two classifiers.
        F1 Score takes into account precision and the recall.
        It is created by finding the the harmonic mean of precision and recall.

        F1 = 2 x (precision x recall)/(precision + recall)
        
        
        
> > ***Precision*** - Precision is the ratio of correctly predicted positive observations to the total predicted positive observations. The question that this metric answer is of all passengers that labeled as survived, how many actually survived? High precision relates to the low false positive rate. We have got 0.788 precision which is pretty good.
> >
> > Precision = TP/TP+FP
> >
> > ***Recall (Sensitivity)*** - Recall is the ratio of correctly predicted positive observations to the all observations in actual class - yes. The question recall answers is: Of all the passengers that truly survived, how many did we label? A recall greater than 0.5 is good.
> >
> > Recall = TP/TP+FN
> >
> > ***F1 score*** - F1 Score is the weighted average of Precision and Recall. Therefore, this score takes both false positives and false negatives into account. Intuitively it is not as easy to understand as accuracy, but F1 is usually more useful than accuracy, especially if you have an uneven class distribution. Accuracy works best if false positives and false negatives have similar cost. If the cost of false positives and false negatives are very different, it’s better to look at both Precision and Recall.
> >
> > F1 Score = 2*(Recall * Precision) / (Recall + Precision)

## **Hyper Parameter optimization (Advanced processing)**
### **Optional for this exercise**

Grid search is an approach to hyperparameter tuning that will methodically build and evaluate a model for each combination of algorithm parameters specified in a grid.

Let’s consider the following example:

Suppose, a machine learning model X takes hyperparameters a1, a2 and a3. In grid searching, you first define the range of values for each of the hyperparameters a1, a2 and a3. You can think of this as an array of values for each of the hyperparameters. Now the grid search technique will construct many versions of X with all the possible combinations of hyperparameter (a1, a2 and a3) values that you defined in the first place. This range of hyperparameter values is referred to as the grid.

Suppose, you defined the grid as:
a1 = [0,1,2,3,4,5]
a2 = [10,20,30,40,5,60]
a3 = [105,105,110,115,120,125]

Note that, the array of values of that you are defining for the hyperparameters has to be legitimate in a sense that you cannot supply Floating type values to the array if the hyperparameter only takes Integer values.

Now, grid search will begin its process of constructing several versions of X with the grid that you just defined.

It will start with the combination of [0,10,105], and it will end with [5,60,125]. It will go through all the intermediate combinations between these two which makes grid search computationally very expensive.
