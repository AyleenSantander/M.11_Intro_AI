```python
# 1. Data Types and Variables
book_title = "Python Programming"
book_quantity = 8
book_price = 29.99
is_available = True

print(f"Book: {book_title}, Quantity: {book_quantity}, Price: ${book_price}, Available: {is_available}")
```

```python
# 2. Lists/# TODO: Add two more books to the list
books = ["Python Basics", "Advanced Python", "Data Science with Python", "Artificial Intelligence", "SQl Fundamental", "Algebra for ML"]
books.append("Machine Learning with Python")
books.insert(0, "Python Crash Course")

print("Updated book list:", books)
```

```python
#Additional Remove book = "SQL Fundamental"
books.remove("SQl Fundamental")
books.sort()
books.insert(3, "Vegan Food is good")
print("Updated book list:", books)
```

```python
# 3. Loops and range()/# TODO: Use a for loop and range() to print books with their indices
print("\nAvailable books:")
for i in range(len(books)):
    print(f"{i}: {books[i]}")
```

```python
# 4. User Input/ Added a new book "EDA book"
new_book = input("Enter a new book title: ").strip()
books.append(new_book)
print("Updated book list:", books)
```

```python
# 5. Functions and Conditional Statements/ # TODO: Check if the book is in the list and return True if available, False otherwise
def check_availability(book_name):
    if book_name in books:
        return True
    else:
        return False

print(f"Is 'Python Basics' available? {check_availability('Python Basics')}")
```

```python
# 6. File I/O/# TODO: Implement a function to read books from a file
def save_books_to_file(filename):
    with open(filename, 'w') as f:
        for book in books:
            f.write(f"{book}\n")

def read_books_from_file(filename):
    with open(filename, 'r') as f:
        return [line.strip() for line in f if line.strip()]

save_books_to_file("books.txt")
print("Books read back from file:", read_books_from_file("books.txt"))
```

```python
# 7. Pickle for serialization/# TODO: Implement a function to load books using pickle
def save_books_pickle(filename):
    with open(filename, 'wb') as f:
        pickle.dump(books, f)

def load_books_pickle(filename):
    with open(filename, 'rb') as f:
        return pickle.load(f)

save_books_pickle("books.pkl")
print("Books loaded from pickle:", load_books_pickle("books.pkl"))
```

```python
#8. SQLite Database Interaction/# TODO: Implement a function to add a book to the SQLite database
def create_books_table():
    conn = sqlite3.connect('library.db')
    cursor = conn.cursor()
    cursor.execute('''CREATE TABLE IF NOT EXISTS books
                      (id INTEGER PRIMARY KEY, title TEXT, quantity INTEGER)''')
    conn.commit()
    conn.close()

def add_book_to_db(title, quantity):
    conn = sqlite3.connect('library.db')
    cursor = conn.cursor()
    cursor.execute("INSERT INTO books (title, quantity) VALUES (?, ?)", (title, quantity))
    conn.commit()
    conn.close()

def get_all_books_from_db():
    conn = sqlite3.connect('library.db')
    cursor = conn.cursor()
    cursor.execute("SELECT id, title, quantity FROM books")
    rows = cursor.fetchall()
    conn.close()
    return rows

create_books_table()
add_book_to_db("Python for Beginners", 10)
for row in get_all_books_from_db():
    print(row)
```

```python
# 9. Web API Interaction (GitHub API)
def get_python_repos():
    url = "https://api.github.com/search/repositories"
    params = {"q": "language:python", "sort": "stars", "order": "desc"}
    response = requests.get(url, params=params)
    if response.status_code != 200:
        print(f"Request failed with status {response.status_code}")
        return []
    data = response.json()
    return [repo['name'] for repo in data['items'][:5]]

print("Top 5 Python repositories on GitHub:", get_python_repos())

print("\nLibrary Management System operations completed.")
```
